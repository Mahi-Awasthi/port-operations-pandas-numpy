# NumPy Array Reshaping Practice

A standalone practice notebook based on the NumPy code cell shared for this project.

## Learning objectives

- Create a one-dimensional NumPy array.
- Reshape an array to specified dimensions and infer a dimension with `-1`.
- Flatten a multidimensional array with `ravel()`.


```python
import numpy as np

values = np.arange(1, 13)
print("Original:", values, values.shape)

matrix_3x4 = values.reshape(3, 4)
print("\n3 x 4:")
print(matrix_3x4)

matrix_2x6 = values.reshape(2, -1)
print("\n2 x inferred columns:")
print(matrix_2x6)

# Flatten a multi-dimensional array
print("\nFlattened:", matrix_3x4.ravel())
```
